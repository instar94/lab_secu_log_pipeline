### 2) 데이터 결함 확인

```python
import numpy as np
```

```python
### 수치형 컬럼 확인
ncol = df.select_dtypes(include=[np.number]).columns
```

- 결측치 확인 : 결함있는 데이터가 있음 
  - Flow Bytes/s
  - Flow Packets/s -> 471개 
  - Flow Duration = 0 인 행 수: 471
  - 동일하게 471개 -> 뭔가 관련이 있을 것으로 보임

```python
inf_counts = np.isinf(df[ncol]).sum()
print("\n[Inf 있는 컬럼]")
print(inf_counts[inf_counts > 0])

nan_counts = df[ncol].isna().sum()
print("\n[NaN 있는 컬럼]")
print(nan_counts[nan_counts > 0])

dup_full = df.duplicated().sum()
print("\n전체 컬럼(Label 포함) 완전 일치 중복 행 수:", dup_full)

neg_counts = (df[ncol] < 0).sum()
print("\n[음수 있는 컬럼]")
print(neg_counts[neg_counts > 0])

zero_dur = df[df["Flow Duration"] == 0]
print("\nFlow Duration = 0 인 행 수:", len(zero_dur))
print(zero_dur["Label"].value_counts())
```

- 음수로 나온 컬럼 확인 

```python
print("Header Length 음수값 분포:")
print(df.loc[df["Fwd Header Length"] < 0, "Fwd Header Length"].describe())

print("\nDuration 음수(17건)와 Packets/s·Bytes/s 음수 겹침:")
neg_dur = df["Flow Duration"] < 0
print("Duration & Packets/s 동시:", (neg_dur & (df["Flow Packets/s"] < 0)).sum())
print("Duration & Bytes/s 동시:", (neg_dur & (df["Flow Bytes/s"] < 0)).sum())

print("\nFlow IAT Min 음수(620건) 중 Duration도 음수인 비율:")
print(df.loc[df["Flow IAT Min"] < 0, "Flow Duration"].lt(0).value_counts())

print("\nmin_seg_size_forward 음수값 분포:")
print(df.loc[df["min_seg_size_forward"] < 0, "min_seg_size_forward"].describe())
```

```python
neg_dur_rows = df[df["Flow Duration"] < 0]
print("Duration<0 행의 Label 분포:")
print(neg_dur_rows["Label"].value_counts())
```

### 3) 데이터 전처리 

```python
# 1. 중복 컬럼 확인 후 정리
same = (df["Fwd Header Length"] == df["Fwd Header Length.1"]).all()
print("Fwd Header Length 중복 컬럼 값 동일:", same)
df = df.drop(columns=["Fwd Header Length.1"])

# 2. Duration=0: Inf/NaN 치환 + 플래그
df["is_zero_duration"] = df["Flow Duration"] == 0
df["Flow Bytes/s"] = df["Flow Bytes/s"].replace([np.inf, -np.inf], 0).fillna(0)
df["Flow Packets/s"] = df["Flow Packets/s"].replace([np.inf, -np.inf], 0)

# 3. Duration<0 행 삭제
before = len(df)
df = df[df["Flow Duration"] >= 0]
print("Duration<0 삭제:", before - len(df))

# 4. Header Length류 극단값 행 삭제
before = len(df)
broken_mask = (
    (df["Fwd Header Length"] < 0)
    | (df["Bwd Header Length"] < 0)
    | (df["min_seg_size_forward"] < 0)
)
df = df[~broken_mask]
print("Header Length 극단값 삭제:", before - len(df))

# 5. 완전 중복 삭제
before = len(df)
df = df.drop_duplicates()
print("완전 중복 삭제:", before - len(df))

# 6. 분석용 컬럼 제거
df = df.drop(columns=["A_source"])

print("\n최종 shape:", df.shape)
```
